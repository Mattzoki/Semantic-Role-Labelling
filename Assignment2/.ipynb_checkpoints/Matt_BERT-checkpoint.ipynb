```python
def extract_gold_from_dataframe(df):
    """
    Extracts the gold labels (target values) for each sentence in the dataframe.

    Args:
        df (pd.DataFrame): A dataframe containing tokenized sentences with a column 'args' for argument labels.

    Returns:
        list: A list of gold labels (target values), with 'O' for non-arguments or predicates, and the actual argument labels for others.
    """
    sentences = df.groupby('sentence_num', sort=False)
    targets = []
    for sentence_num, sentence in sentences:
        print(f"\rProgress:{sentence_num}", end="")
        for _, row in sentence.iterrows():
            if row['args'] == '_' or row['args'] == 'V':
                targets.append('O')
        # To seperate different progress bars for diff files
        print()
```

```python
def create_label_mapping(df, column='args'):
    """
    Creates a label mapping for the specified column in the dataframe, encoding labels as integers.

    Args:
        df (pd.DataFrame): The dataframe containing the data.
        column (str): The column name to extract and map labels from (default is 'args').

    Returns:
        dict: A dictionary mapping the original labels to their encoded integer values.
    """
    df = df.copy()
    df[column] = df[column].replace({'_': 'O', 'V': 'O'})
    unique_labels = df[column].unique()
    encoder = LabelEncoder()
    labels_encoded = encoder.fit_transform(unique_labels)
    label_mapping = dict(zip(encoder.classes_, labels_encoded))

    if 'O' not in label_mapping:
        label_mapping['O'] = max(labels_encoded) + 1

    return label_mapping
```

```python
def convert_df(df):
    """
    Converts a dataframe into a list of dictionaries, with tokens, labels, and predicates for each sentence.

    Args:
        df (pd.DataFrame): The dataframe containing tokenized sentences and their corresponding labels.

    Returns:
        list: A list of dictionaries, where each dictionary contains 'tokens', 'labels', and 'predicate' for a sentence.
    """
    label_mapping = create_label_mapping(train_df, column='args')
    result = [
        {
            'tokens': group['token'].tolist(),
            'labels': [label_mapping.get(item, label_mapping['O']) for item in group['args']],
            'predicate': [0 if val == 0 else 1 for val in group['chosen_predicate']]
        }
        for _, group in df.groupby('sentence_num', sort=False)
    ]
    return result
```

```python
def augment_data(data):
    """
    Augments the input data by adding a special token before the predicate and modifying labels accordingly.

    Args:
        data (list): A list of dictionaries, where each dictionary represents a sentence with 'tokens', 'labels', and 'predicate'.

    Returns:
        list: A list of dictionaries, where each dictionary contains the augmented 'tokens' for a sentence.
    """
    final = []
    for sentence in data:

        tokens = sentence['tokens']
        predicates = sentence['predicate']
        labels = sentence['labels']

        processed_tokens = []
        processed_labels = []
        processed_predicates = []
        for idx, (token, predicate) in enumerate(zip(tokens, predicates)):
            if predicate == 1:
                processed_labels.append(-100)
                processed_predicates.append(predicates[idx])
                processed_tokens.append("[PRED]")

            processed_tokens.append(token)
            processed_labels.append(labels[idx])
            processed_predicates.append(predicates[idx])

        final.append({
            'tokens': processed_tokens,
            'labels': processed_labels,
            'predicates': processed_predicates
        })
    return final
```

```python
def tokenized_input(data):
    """
    Tokenizes input sentences and aligns their labels with the tokenized tokens.

    Args:
        data (list): A list of dictionaries, where each dictionary represents a sentence with 'tokens' and 'labels'.

    Returns:
        list: A list of dictionaries, where each dictionary contains the tokenized input, aligned labels, and additional tokenization information.
    """
    tokenized_data = []
    for sent in data:
        single_sent = sent.copy()
        tokenized_input = tokenizer(sent["tokens"], is_split_into_words=True)
        tokens = tokenizer.convert_ids_to_tokens(tokenized_input["input_ids"])
        word_ids = tokenized_input.word_ids()
        aligned_labels = [-100 if i is None else sent["labels"][i]
                          for i in word_ids]

        single_sent["labels"] = aligned_labels
        single_sent["input_ids"] = tokenized_input["input_ids"]
        single_sent["attention_mask"] = tokenized_input["attention_mask"]
        single_sent['word_ids'] = word_ids
        single_sent['tokenized_tokens'] = tokens
        tokenized_data.append(single_sent)
    return tokenized_data
```

```python
def compute_metrics(p):
    """
    Computes the evaluation metrics (precision, recall, F1, accuracy) for the model's predictions.

    Args:
        p (tuple): A tuple containing the model's predictions and the true labels.

    Returns:
        dict: A dictionary containing the precision, recall, F1 score, and accuracy.
    """
    predictions, labels = p
    predictions = np.argmax(predictions, axis=2)

    true_predictions = [
        [label_list[p] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]
    true_labels = [
        [label_list[l] for (p, l) in zip(prediction, label) if l != -100]
        for prediction, label in zip(predictions, labels)
    ]

    results = metric.compute(
        predictions=true_predictions, references=true_labels)
    return {
        "precision": results["overall_precision"],
        "recall": results["overall_recall"],
        "f1": results["overall_f1"],
        "accuracy": results["overall_accuracy"],
    }
```

```python
def make_dataset(processed_doc):
    """
    Converts a list of processed examples into a Hugging Face Dataset that is suitable for BERT.

    Args:
        processed_doc (list): A list of dictionaries, each containing 'input_ids', 'attention_mask', and 'labels'.

    Returns:
        Dataset: A Hugging Face Dataset
    """
    final_dataset = Dataset.from_dict({
        'input_ids': [example['input_ids'] for example in processed_doc],
        'attention_mask': [example['attention_mask'] for example in processed_doc],
        'labels': [example['labels'] for example in processed_doc]
    })
    return final_dataset


label_list = list(label_mapping.keys())
train_dataset = make_dataset(processed_train)
dev_dataset = make_dataset(processed_dev)
test_dataset = make_dataset(processed_test)
```

```python
def process_data(data_list):
    """
    Processes a list of data dictionaries, where each dictionary contains tokenized tokens,
    word IDs, predictions, and gold labels. It filters out unwanted tokens and aggregates 
    predictions and gold labels based on word ID groups.

    Args:
        data_list (list): A list of dictionaries.

    Returns:
        tuple: A tuple containing two lists:
            - processed_predictions_list: A list of processed prediction labels for each example.
            - processed_gold_list: A list of processed gold labels for each example.
    """
    def process_single(data):
        """
        Processes a single data dictionary by filtering out unwanted tokens and aggregating
        predictions and gold labels based on word ID groups.

        Args:
            data (dict): A dictionary'.

        Returns:
            tuple: A tuple containing the processed predictions and gold labels for the input data.
        """
        word_ids = data['word_ids']
        predictions = data['predictions']
        gold = data['gold']
        tokenized_tokens = data['tokenized_tokens']

        word_ids = [b for a, b in zip(tokenized_tokens, word_ids) if a not in [
            '[CLS]', '[SEP]', '[PAD]', '[PRED]', '[UNK]', '[MASK]']]
        predictions = [b for a, b in zip(tokenized_tokens, predictions) if a not in [
            '[CLS]', '[SEP]', '[PAD]', '[PRED]', '[UNK]', '[MASK]']]
        gold = [b for a, b in zip(tokenized_tokens, gold) if a not in [
            '[CLS]', '[SEP]', '[PAD]', '[PRED]', '[UNK]', '[MASK]']]

        processed_predictions = []
        processed_gold = []

        current_word_id = None
        current_predictions = []
        current_gold = []

        for idx, word_id in enumerate(word_ids):
            if word_id != current_word_id:
                if current_predictions:
                    processed_predictions.append(
                        Counter(current_predictions).most_common(1)[0][0])
                    processed_gold.append(
                        Counter(current_gold).most_common(1)[0][0])

                current_word_id = word_id
                current_predictions = [predictions[idx]]
                current_gold = [gold[idx]]
            else:
                current_predictions.append(predictions[idx])
                current_gold.append(gold[idx])

        if current_predictions:
            processed_predictions.append(
                Counter(current_predictions).most_common(1)[0][0])
            processed_gold.append(Counter(current_gold).most_common(1)[0][0])

        return processed_predictions, processed_gold

    processed_predictions_list = []
    processed_gold_list = []

    for data in data_list:
        processed_predictions, processed_gold = process_single(data)
        processed_predictions_list.append(processed_predictions)
        processed_gold_list.append(processed_gold)

    return processed_predictions_list, processed_gold_list
```

```python
def output_tsv(input_file, output_file, predictions, gold_labels, id_to_label):
    """
    Writes predictions and gold labels to an output TSV file, aligning them with the tokens
    in the input file. The labels are mapped using the provided `id_to_label` dictionary.

    Args:
        input_file (str): Path to the input file containing tokens.
        output_file (str): Path to the output file where results will be written.
        predictions (list): A list of predicted labels
        gold_labels (list): A list of gold labels
        id_to_label (dict): A dictionary mapping label IDs to label names.

    Returns:
        None: The function writes output to the specified `output_file`.
    """
    if isinstance(predictions[0], list):
        predictions = [label for sentence in predictions for label in sentence]
    if isinstance(gold_labels[0], list):
        gold_labels = [label for sentence in gold_labels for label in sentence]

    with open(output_file, "w") as outfile:
        index = 0

        with open(input_file, "r") as infile:
            for line in infile:
                if len(line.rstrip("\n").split()) > 0:
                    token = line.split()[3]

                    pred_label = id_to_label.get(int(predictions[index]), "O")
                    gold_label = id_to_label.get(int(gold_labels[index]), "O")

                    outfile.write(line.rstrip("\n") +
                                  f"\t{pred_label}\t{gold_label}\n")
                    index += 1

    print(f"Output written to {output_file}")
```

```python
def evaluate_file(tsv_file, id_to_label):
    """
    Evaluates the model's predictions by comparing them with gold labels in a TSV file.
    It computes the classification report and confusion matrix for the predicted labels,
    and visualizes the confusion matrix.

    Args:
        tsv_file (str): Path to the TSV file containing tokens, gold labels, and predictions.
        id_to_label (dict): A mapping from label IDs to actual labels.

    Returns:
        None: The function prints the classification report and displays the confusion matrix plot.
    """
    columns = [
        'sentence_num', 'token', 'lemma', 'uni_pos', 'penn_pos', 'morph_notes', 'head',
        'dep', 'enhanced_dep', 'notes', 'predicate', 'args', 'chosen_predicate', 'pred', 'gold'
    ]

    df = pd.read_csv(tsv_file, sep="\t", header=None, names=columns)

    def map_label(value):
        """
        Maps a given label to its corresponding label using the `id_to_label` dictionary.

        Args:
            value: The label value to be mapped.

        Returns:
            str: The mapped label or the original label if it's already a string.
        """
        value = str(value).strip()
        if value.isdigit():
            return id_to_label.get(int(value), "O")
        return value

    y_true = df["gold"].apply(map_label).tolist()
    y_pred = df["pred"].apply(map_label).tolist()

    unique_labels = sorted(set(y_true) | set(y_pred))

    report = classification_report(
        y_true, y_pred, labels=unique_labels, digits=3, zero_division=0)
    print("Classification Report:")
    print(report)

    cf_matrix = confusion_matrix(y_true, y_pred, labels=unique_labels)

    # We diffuse the value of O marked as O so that the confusion matrix is clearer
    if 'O' in unique_labels:
        o_index = unique_labels.index("O")
        cf_matrix[o_index, o_index] /= 100

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cf_matrix, display_labels=unique_labels)

    fig, ax = plt.subplots(figsize=(18, 18))
    disp.plot(cmap="cividis", xticks_rotation=45, ax=ax)

    plt.title("Confusion Matrix", fontsize=14)
    plt.xlabel("Predicted Labels", fontsize=12)
    plt.ylabel("True Labels", fontsize=12)

    plt.show()
```

The easiest roles to identify are those with higher support, such as ARG0 (subject), ARG1 (object), and ARGM-NEG (negation), which achieved F1 scores of 0.867, 0.873, and 0.953, respectively. These roles are more frequent and are typically easier for the model to detect due to their clearer syntactic and semantic features. Roles like ARG0 and ARG1 also benefit from a larger number of training examples, aiding the model in learning their patterns effectively.

On the other hand, some roles were more challenging. For instance, ARG3, ARG5, C-ARG0, and R-ARGM roles had very low or zero recall and F1 scores, indicating that the model struggled to identify these semantic roles effectively. This may be due to their lower frequency in the dataset or their more complex syntactic patterns. Additionally, roles such as ARGM-PRD (predicate arguments) and C-ARG1 (complex arguments) also exhibited weaker performance, reflecting the model’s difficulty with capturing less frequent or more nuanced semantic relationships.

The macro average scores (precision: 0.471, recall: 0.443, F1: 0.446) suggest that, despite the high accuracy, the model may have performance imbalances across different labels.

```python
def main_preprocessor(sentence_data, BERTtokenizer):
    """
    Preprocesses a DataFrame containing sentence data to prepare it for input into a BERT-based model. It handles tokenization,
    token classification (by augmenting predicates with "[PRED]"), and returns the tokenized data along with a final dataset for BERT input.

    Parameters:
    sentence_data (DataFrame): A dataframe containing columns 'token' and 'chosen_predicate', where 'token' represents the words 
                                in the sentence and 'chosen_predicate' is a binary column indicating whether a token is a predicate (1) or not (0).
    BERTtokenizer (PreTrainedTokenizer): A pre-trained BERT tokenizer used to tokenize the sentence data and prepare it for model input.

    Returns:
    tuple: A tuple containing:
        - tokenized_data (list): A list of dictionaries, where each dictionary contains:
            - 'input_ids' (list): List of token IDs for each tokenized sentence.
            - 'attention_mask' (list): Attention mask indicating which tokens are actual words and which are padding.
            - 'word_ids' (list): List of word IDs, mapping each token to its original word.
            - 'tokenized_tokens' (list): List of tokenized tokens corresponding to the input sentence.
        - final_dataset (Dataset): A HuggingFace Dataset containing the 'input_ids' and 'attention_mask' for use with BERT.
    """

    x = [
        {
            'tokens': sentence_data['token'].tolist(),
            'predicate': [0 if val == 0 else 1 for val in sentence_data['chosen_predicate']]
        }
    ]

    y = []
    for sentence in x:
        tokens = sentence['tokens']
        predicates = sentence['predicate']

        processed_tokens = []
        processed_predicates = []
        for idx, (token, predicate) in enumerate(zip(tokens, predicates)):
            if predicate == 1:
                processed_predicates.append(predicates[idx])
                processed_tokens.append("[PRED]")

            processed_tokens.append(token)
            processed_predicates.append(predicates[idx])

        y.append({
            'tokens': processed_tokens,
            'predicates': processed_predicates
        })

    tokenized_data = []

    for sent in y:
        single_sent = {}
        tokenized_input = BERTtokenizer(
            sent["tokens"], is_split_into_words=True)
        tokens = BERTtokenizer.convert_ids_to_tokens(
            tokenized_input["input_ids"])
        word_ids = tokenized_input.word_ids()

        single_sent["input_ids"] = tokenized_input["input_ids"]
        single_sent["attention_mask"] = tokenized_input["attention_mask"]
        single_sent['word_ids'] = word_ids
        single_sent['tokenized_tokens'] = tokens
        tokenized_data.append(single_sent)

    processed_data = []
    for example in tokenized_data:
        processed_data.append({
            "input_ids": tokenized_data[0]["input_ids"],
            "attention_mask": tokenized_data[0]["attention_mask"],
        })

    final_dataset = Dataset.from_dict({
        'input_ids': [example['input_ids'] for example in processed_data],
        'attention_mask': [example['attention_mask'] for example in processed_data]
    })

    return tokenized_data, final_dataset
```

```python
def split_predicates_flat(sent, sent_target):
    # Initialize the lists to store the results
    sentence_duplicates = []
    target_duplicates = []

    # Iterate through each word in the sentence
    for idx, (word, target) in enumerate(zip(sent, sent_target)):
        # If the word is a predicate (target == 1)
        if target == 1:
            # Create a copy of the sentence and target list
            sentence_copy = sent[:]
            target_copy = sent_target[:]

            # Set all predicates (other than the current one) to 0
            for i in range(len(target_copy)):
                if i != idx:
                    target_copy[i] = 0

            # Append the modified sentence and target list to the output lists
            sentence_duplicates.extend(sentence_copy)
            target_duplicates.extend(target_copy)

    return sentence_duplicates, target_duplicates
```

```python
def transform_to_test_sentence(sent, target, BERTtrainer, BERTmodel, BERTtokenizer):
    """
    Transforms a sentence and its target into tokenized input suitable for semantic role labeling (SRL) using a BERT-based model. 
    It processes the input sentence, predicts semantic roles for each token, and returns the predicted roles.

    Parameters:
    sent (list): A list of tokens (words) in the sentence.
    target (list): A list of binary values indicating the presence of a (predicate). The length must match the length of `sent`.
    BERTtrainer (Trainer): A HuggingFace Trainer used for making predictions on the processed sentence.
    BERTmodel (PreTrainedModel): A pre-trained BERT model for semantic role labeling.
    BERTtokenizer (PreTrainedTokenizer): A pre-trained BERT tokenizer for tokenizing the sentence.

    Returns:
    list: A list of predicted semantic roles for each word in the input sentence. 
    The roles are returned as labels mapped from integers to their corresponding semantic roles.
    """

    sent, target = split_predicates_flat(sent, target)

    if len(sent) != len(target):
        raise ValueError(
            "The length of the sentence and the array to mark predicate must be the same.")

    label_to_id = {'ARG0': 45, 'ARG1': 0, 'ARG1-DSP': 1, 'ARG2': 17, 'ARG3': 20, 'ARG4': 16,
                   'ARG5': 26, 'ARGA': 3, 'ARGM-ADJ': 9, 'ARGM-ADV': 19, 'ARGM-CAU': 18, 'ARGM-COM': 24,
                   'ARGM-CXN': 5, 'ARGM-DIR': 21, 'ARGM-DIS': 15, 'ARGM-EXT': 46, 'ARGM-GOL': 8,
                   'ARGM-LOC': 47, 'ARGM-LVB': 44, 'ARGM-MNR': 4, 'ARGM-MOD': 11, 'ARGM-NEG': 14,
                   'ARGM-PRD': 10, 'ARGM-PRP': 23, 'ARGM-PRR': 13, 'ARGM-REC': 12, 'ARGM-TMP': 28,
                   'C-ARG0': 22, 'C-ARG1': 51, 'C-ARG1-DSP': 58, 'C-ARG2': 30, 'C-ARG3': 25, 'C-ARG4': 56,
                   'C-ARGM-ADV': 48, 'C-ARGM-COM': 35, 'C-ARGM-CXN': 2, 'C-ARGM-DIR': 29, 'C-ARGM-EXT': 52,
                   'C-ARGM-GOL': 57, 'C-ARGM-LOC': 54, 'C-ARGM-MNR': 40, 'C-ARGM-PRP': 39, 'C-ARGM-PRR': 32,
                   'C-ARGM-TMP': 31, 'C-V': 49, 'O': 27, 'R-ARG0': 43, 'R-ARG1': 55, 'R-ARG2': 33, 'R-ARG3': 37,
                   'R-ARG4': 50, 'R-ARGM-ADV': 34, 'R-ARGM-CAU': 36, 'R-ARGM-COM': 53, 'R-ARGM-DIR': 6,
                   'R-ARGM-GOL': 7, 'R-ARGM-LOC': 42, 'R-ARGM-MNR': 38, 'R-ARGM-TMP': 41}
    id_to_label = {v: k for k, v in label_to_id.items()}

    test_sentence = []
    for token, label in zip(sent, target):
        test_sentence.append(
            {"token": token, "chosen_predicate": token if label == 1 else 0})

    sentence_data = pd.DataFrame(test_sentence)
    tokenized_data, final = main_preprocessor(sentence_data, BERTtokenizer)

    predictions, _, _ = BERTtrainer.predict(final)
    predictions = np.argmax(predictions, axis=2)

    for idx, i in enumerate(tokenized_data):
        length = len(i['word_ids'])
        i['predictions'] = predictions[idx][:length]

    processed_predictions = []
    for data in tokenized_data:
        word_ids = data['word_ids']
        predictions = data['predictions']

        word_ids = [b for a, b in zip(data['tokenized_tokens'], word_ids) if a not in [
            '[CLS]', '[SEP]', '[PAD]', '[PRED]', '[UNK]', '[MASK]']]
        predictions = [b for a, b in zip(data['tokenized_tokens'], predictions) if a not in [
            '[CLS]', '[SEP]', '[PAD]', '[PRED]', '[UNK]', '[MASK]']]

        current_word_id = None
        current_predictions = []

        for idx, word_id in enumerate(word_ids):
            if word_id != current_word_id:
                if current_predictions:
                    predicted_label_id = Counter(
                        current_predictions).most_common(1)[0][0]
                    processed_predictions.append(
                        id_to_label[predicted_label_id])

                current_word_id = word_id
                current_predictions = [predictions[idx]]
            else:
                current_predictions.append(predictions[idx])

        if current_predictions:
            predicted_label_id = Counter(
                current_predictions).most_common(1)[0][0]
            processed_predictions.append(
                id_to_label[predicted_label_id])

    return sent, target, processed_predictions


sent = ["The", "dog", "barks", "and", "the",
        "woman", "ran", "away", "quickly", "."]
sent_target = [0, 0, 1, 0, 0, 0, 1, 0, 0, 0]


# Example usage with model and tokenizer
processed_sent, processed_target, sem_role,  = transform_to_test_sentence(
    sent, sent_target, BERTtrainer, BERTmodel, BERTtokenizer)

# Print headers
print(f"{'Token':<15}{'Role':<15}{'Target':<15}")

# Iterate through the lists and print each corresponding element
for token, role, target in zip(processed_sent, sem_role, processed_target):
    print(f"{token:<15}{role:<15}{target:<15}")
```

```python
def get_logreg_bert_predictions_df(
    json_file, loaded_logreg_model, loaded_vectorizer, tokenizer, model, sorted_unique_labels
):
    with open(json_file, "r") as infile:
        list_dicts = json.load(infile)

    all_rows = []

    for s_idx, sentence in enumerate(list_dicts):
        tokens = sentence['sent_tokens']
        gold_labels = sentence['sent_gold']

        predictions_on_standalone_logreg = adv_utils.stand_alone_sent_logreg(
            sentence['sent_tokens'], sentence['sent_predicate'], loaded_logreg_model, loaded_vectorizer, loaded_encoder
        )

        predictions_on_standalone_bert = adv_utils.stand_alone_sent_bert(
            sentence['sent_tokens'], sentence['sent_predicate'], model, tokenizer
        )
        for idx, (token, logreg_prediction, bert_prediction, gold_label) in enumerate(
            zip(tokens, predictions_on_standalone_logreg,
                predictions_on_standalone_bert, gold_labels)
        ):

            all_rows.append([
                f'{s_idx+1}',
                token if not token.endswith(
                    '[PRED]') else token.replace('[PRED]', ''),
                logreg_prediction,
                bert_prediction,
                gold_label
            ])

    df_predictions = pd.DataFrame(
        all_rows,
        columns=['Sentence_idx', 'Token', 'LogReg Prediction',
                 'BERT Prediction', 'Gold Label']
    )

    return df_predictions
```

```python
predictions = get_logreg_bert_predictions_df(
    "./one.json", loaded_logreg_model, loaded_vectorizer, tokenizer, model, sorted_unique_labels
)
```
